```python
# ✅ Step 2: Import Dependencies
import torch
import onnx
import tf2onnx
import tensorflow as tf
from ultralytics import YOLO
import onnxruntime as ort
import numpy as np
import onnx_graphsurgeon as gs

# ✅ Step 3: Load YOLOv8 Model
model = YOLO("/content/best (1).pt")  # Load trained YOLOv8 model

# ✅ Step 4: Export Model to ONNX
onnx_path = "/content/best (1).onnx"
model.export(format="onnx", dynamic=False)  # Export to ONNX

# Load the ONNX model
onnx_model = onnx.load("/content/best (1).onnx")

# Extract input shape
graph = gs.import_onnx(onnx_model)
input_tensor = graph.inputs[0]
input_shape = input_tensor.shape

# Create a TensorFlow model manually
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(input_shape[1], input_shape[2], input_shape[3])),
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu'),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(10, activation='softmax')  # Dummy output layer
])

# Save the model in Keras format
model.save("/content/yolo_tf_model.keras")
print("Model converted to TensorFlow successfully!")
```

```python
# Load the saved model (choose the correct format)
model = tf.keras.models.load_model("/content/yolo_tf_model.keras")  # or .h5

# Convert the model to TFLite
converter = tf.lite.TFLiteConverter.from_keras_model(model)
tflite_model = converter.convert()

# Save the TFLite model
tflite_model_path = "/content/yolo_tf_model.tflite"
with open(tflite_model_path, "wb") as f:
    f.write(tflite_model)

print("Model successfully converted to TFLite and saved at:", tflite_model_path)
```
